```python

import random
import math

N = 8

# 1. Calculate number of attacking queen pairs
def calculate_conflicts(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):

            # Same column or diagonal
            if board[i] == board[j]:
                conflicts += 1

            elif abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# 2. Display chessboard
def display_board(board):
    print("\nChessboard:")

    for row in range(N):
        for col in range(N):
            if board[row] == col:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


# 3. Simulated Annealing Algorithm
def simulated_annealing():

    # Random initial state
    current = [random.randint(0, N - 1) for _ in range(N)]

    current_conflicts = calculate_conflicts(current)

    best = current.copy()
    best_conflicts = current_conflicts

    temperature = 100
    cooling_rate = 0.99
    min_temperature = 0.001

    iteration = 0

    print("Initial Board:", current)
    print("Initial Conflicts:", current_conflicts)

    while temperature > min_temperature:

        # Generate neighboring state
        neighbor = current.copy()

        row = random.randint(0, N - 1)
        new_col = random.randint(0, N - 1)

        neighbor[row] = new_col

        neighbor_conflicts = calculate_conflicts(neighbor)

        # Calculate difference
        delta = neighbor_conflicts - current_conflicts

        # Accept better solution
        if delta < 0:
            current = neighbor
            current_conflicts = neighbor_conflicts

        # Sometimes accept worse solution
        else:
            probability = math.exp(-delta / temperature)

            if random.random() < probability:
                current = neighbor
                current_conflicts = neighbor_conflicts

        # Update best solution
        if current_conflicts < best_conflicts:
            best = current.copy()
            best_conflicts = current_conflicts

        iteration += 1

        # Stop when solution is found
        if best_conflicts == 0:
            break

        # Cool down temperature
        temperature *= cooling_rate

    print("\nTotal Iterations:", iteration)

    return best, best_conflicts


# 4. RUN PROGRAM
random.seed(42)

solution, conflicts = simulated_annealing()

# 5. DISPLAY RESULT
print("\n------ FINAL RESULT ------")

print("Final Queen Positions:", solution)
print("Number of Conflicts:", conflicts)

if conflicts == 0:
    print("Solution Found! All 8 Queens are Safe.")
else:
    print("Solution Not Found. Try Running Again.")

display_board(solution)
```
