```python

import random
import numpy as np

# 1. INPUT
NUM_TASKS = int(input("Enter number of tasks: "))
NUM_VMS = int(input("Enter number of Virtual Machines: "))

VM_SPEEDS = []

for i in range(NUM_VMS):
    speed = int(input(f"Enter speed of VM {i} (MIPS): "))
    VM_SPEEDS.append(speed)

TASK_WORKLOADS = []

for i in range(NUM_TASKS):
    workload = int(input(f"Enter workload of Task {i}: "))
    TASK_WORKLOADS.append(workload)

# 2. EXECUTION TIME MATRIX
EXEC_MATRIX = np.zeros((NUM_TASKS, NUM_VMS))

for t in range(NUM_TASKS):
    for v in range(NUM_VMS):
        EXEC_MATRIX[t][v] = TASK_WORKLOADS[t] / VM_SPEEDS[v]


# 3. CALCULATE MAKESPAN
def calculate_makespan(schedule):
    vm_loads = np.zeros(NUM_VMS)

    for task_id, vm_id in enumerate(schedule):
        vm_loads[vm_id] += EXEC_MATRIX[task_id][vm_id]

    return np.max(vm_loads)


# 4. HILL CLIMBING ALGORITHM
def hill_climbing():

    # Generate random initial schedule
    current = [
        random.randint(0, NUM_VMS - 1)
        for _ in range(NUM_TASKS)
    ]

    current_makespan = calculate_makespan(current)

    print("\nInitial Schedule:", current)
    print("Initial Makespan:", round(current_makespan, 2))

    iteration = 0

    while True:
        best_neighbor = current.copy()
        best_neighbor_makespan = current_makespan

        # Generate neighboring solutions
        for task_id in range(NUM_TASKS):
            for vm_id in range(NUM_VMS):

                neighbor = current.copy()
                neighbor[task_id] = vm_id

                neighbor_makespan = calculate_makespan(neighbor)

                # Check for improvement
                if neighbor_makespan < best_neighbor_makespan:
                    best_neighbor = neighbor
                    best_neighbor_makespan = neighbor_makespan

        # Stop if no better neighbor is found
        if best_neighbor_makespan >= current_makespan:
            break

        current = best_neighbor
        current_makespan = best_neighbor_makespan

        iteration += 1

        print(
            f"Iteration {iteration}: "
            f"Makespan = {current_makespan:.2f} seconds"
        )

    return current, current_makespan


# 5. RUN ALGORITHM
random.seed(101)

initial_schedule = [
    random.randint(0, NUM_VMS - 1)
    for _ in range(NUM_TASKS)
]

initial_makespan = calculate_makespan(initial_schedule)

best_schedule, best_makespan = hill_climbing()

# 6. FINAL RESULTS
print("\n------ OPTIMIZATION COMPLETE ------")

print("Best Task Schedule:", best_schedule)

print(f"Initial Makespan: {initial_makespan:.2f} seconds")
print(f"Optimized Makespan: {best_makespan:.2f} seconds")

print(
    f"Time Saved: "
    f"{initial_makespan - best_makespan:.2f} seconds"
)

improvement = (
    (initial_makespan - best_makespan)
    / initial_makespan
) * 100

print(f"Improvement: {improvement:.2f}%")

# 7. VM LOAD BALANCING
print("\n------ VM RESOURCE LOAD ------")

vm_loads = np.zeros(NUM_VMS)

for task_id, vm_id in enumerate(best_schedule):
    vm_loads[vm_id] += EXEC_MATRIX[task_id][vm_id]

for vm_id in range(NUM_VMS):
    print(
        f"VM {vm_id} "
        f"(Speed: {VM_SPEEDS[vm_id]} MIPS) "
        f"-> Processing Time: {vm_loads[vm_id]:.2f} seconds"
    )
2
```
