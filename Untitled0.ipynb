```python
import random
import numpy as np

NUM_TASKS = 20
NUM_VMS = 4
VM_SPEEDS = [100, 200, 300, 500]

random.seed(101)
TASK_WORKLOADS = [random.randint(500, 4000) for _ in range(NUM_TASKS)]

POP_SIZE = 30
GENERATIONS = 100
CROSSOVER_RATE = 0.85
INITIAL_MUTATION_RATE = 0.20

EXEC_MATRIX = np.zeros((NUM_TASKS, NUM_VMS))
for t in range(NUM_TASKS):
    for v in range(NUM_VMS):
        EXEC_MATRIX[t][v] = TASK_WORKLOADS[t] / VM_SPEEDS[v]

def calculate_makespan(chromosome):
    vm_loads = np.zeros(NUM_VMS)
    for task_id, vm_id in enumerate(chromosome):
        vm_loads[vm_id] += EXEC_MATRIX[task_id][vm_id]
    return np.max(vm_loads)

def fitness_function(chromosome):
    return 1.0 / calculate_makespan(chromosome)

def create_initial_population():
    return [[random.randint(0, NUM_VMS - 1) for _ in range(NUM_TASKS)] for _ in range(POP_SIZE)]

def tournament_selection(population, fitnesses, k=3):
    selected_indices = random.sample(range(len(population)), k)
    best_index = max(selected_indices, key=lambda idx: fitnesses[idx])
    return population[best_index]

def single_point_crossover(parent1, parent2):
    if random.random() < CROSSOVER_RATE:
        point = random.randint(1, NUM_TASKS - 1)
        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]
        return child1, child2
    return parent1.copy(), parent2.copy()

def mutate(chromosome, current_mutation_rate):
    for task_id in range(NUM_TASKS):
        if random.random() < current_mutation_rate:
            chromosome[task_id] = random.randint(0, NUM_VMS - 1)
    return chromosome

population = create_initial_population()

initial_fitnesses = [fitness_function(ind) for ind in population]
initial_best_idx = np.argmax(initial_fitnesses)
initial_best_schedule = population[initial_best_idx]
initial_makespan = calculate_makespan(initial_best_schedule)

print(f"--- Evolution Started ---")
print(f"Initial Random Setup Makespan: {initial_makespan:.2f} seconds\n")

for generation in range(GENERATIONS):
    fitnesses = [fitness_function(ind) for ind in population]

    best_idx = np.argmax(fitnesses)
    elite = population[best_idx].copy()

    current_mutation_rate = max(0.02, INITIAL_MUTATION_RATE * (1.0 - generation / GENERATIONS))

    new_population = []
    new_population.append(elite)

    while len(new_population) < POP_SIZE:
        p1 = tournament_selection(population, fitnesses)
        p2 = tournament_selection(population, fitnesses)

        c1, c2 = single_point_crossover(p1, p2)

        new_population.append(mutate(c1, current_mutation_rate))
        if len(new_population) < POP_SIZE:
            new_population.append(mutate(c2, current_mutation_rate))

    population = new_population

final_fitnesses = [fitness_function(ind) for ind in population]
final_best_idx = np.argmax(final_fitnesses)
final_best_schedule = population[final_best_idx]
final_makespan = calculate_makespan(final_best_schedule)

print(f"--- Optimization Complete ---")
print(f"Optimized Best Makespan: {final_makespan:.2f} seconds")
print(f"Total Time Saved: {initial_makespan - final_makespan:.2f} seconds (Improvement of {((initial_makespan - final_makespan)/initial_makespan)*100:.1f}%)")

print("\nFinal Resource Load Balancing Matrix:")
vm_final_loads = np.zeros(NUM_VMS)
for task_id, vm_id in enumerate(final_best_schedule):
    vm_final_loads[vm_id] += EXEC_MATRIX[task_id][vm_id]

for vm_id in range(NUM_VMS):
    print(f"  VM {vm_id} (Speed: {VM_SPEEDS[vm_id]} MIPS) -> Active Processing Time: {vm_final_loads[vm_id]:.2f} seconds")
```
