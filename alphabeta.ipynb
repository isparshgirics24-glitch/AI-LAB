```python
import ast
import math

pruned_branches = []

def alpha_beta_search(tree, root_name="A", is_max=True):
    pruned_branches.clear()
    if is_max:
        return max_value(tree, -math.inf, math.inf, root_name)
    else:
        return min_value(tree, -math.inf, math.inf, root_name)

def max_value(node, alpha, beta, path_name):
    # TERMINAL-TEST(state): If node is a number, it is a leaf
    if isinstance(node, (int, float)):
        return node, [f"{path_name}({node})"]

    v = -math.inf
    best_path = []

    for i, child in enumerate(node):
        child_name = f"{path_name}.{i+1}"
        val, path = min_value(child, alpha, beta, child_name)

        if val > v:
            v = val
            best_path = [path_name] + path

        # Beta cutoff: if v >= beta then return v
        if v >= beta:
            for j in range(i + 1, len(node)):
                pruned_branches.append(f"{path_name}.{j+1} -> {node[j]}")
            return v, best_path

        alpha = max(alpha, v)

    return v, best_path

def min_value(node, alpha, beta, path_name):
    # TERMINAL-TEST(state): If node is a number, it is a leaf
    if isinstance(node, (int, float)):
        return node, [f"{path_name}({node})"]

    v = math.inf
    best_path = []

    for i, child in enumerate(node):
        child_name = f"{path_name}.{i+1}"
        val, path = max_value(child, alpha, beta, child_name)

        if val < v:
            v = val
            best_path = [path_name] + path

        # Alpha cutoff: if v <= alpha then return v
        if v <= alpha:
            for j in range(i + 1, len(node)):
                pruned_branches.append(f"{path_name}.{j+1} -> {node[j]}")
            return v, best_path

        beta = min(beta, v)

    return v, best_path

# --- Main Execution ---
if __name__ == "__main__":
    root_name = input("Enter root node name (e.g., A or Root): ").strip() or "Root"
    root_type = input("Is the root node MAX or MIN? (MAX/MIN): ").strip().upper()
    is_max = (root_type != "MIN")

    print("\nEnter the game tree as nested lists:")
    print("Example (4-level tree): [[[[5, 6], [7, 4, 5]], [[3]]], [[[6], [6, 9]], [[7]]], [[[5]], [[9, 8], [6]]]]")
    tree_input = input("Tree: ")
    game_tree = ast.literal_eval(tree_input)

    root_val, opt_path = alpha_beta_search(game_tree, root_name, is_max)

    print("\n--- Alpha-Beta Pruning Results ---")
    print(f"Value of Root Node {root_name} ({'MAX' if is_max else 'MIN'}): {root_val}")
    print(f"Optimal Path: {' -> '.join(opt_path)}")
    print(f"Pruned (Cutoff) Branches: {pruned_branches}")
```
